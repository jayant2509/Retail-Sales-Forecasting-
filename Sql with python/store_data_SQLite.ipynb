```python

conn = sqlite3.connect("Sales.db")


df = pd.read_sql_query("SELECT * FROM Sales", conn)


print(df.columns)


df_customers = df[['Customer ID', 'Customer Name', 'Segment', 'Region']].drop_duplicates()

# Extract Products table (drop duplicates)
df_products = df[['Product ID', 'Category', 'Sub-Category', 'Product Name']].drop_duplicates()

# Extract Orders table
df_orders = df[['Order ID', 'Order Date', 'Customer ID', 'Product ID', 'Sales']]

# Create new tables in DB (replace existing if any)
df_customers.to_sql('Customers', conn, if_exists='replace', index=False)
df_products.to_sql('Products', conn, if_exists='replace', index=False)
df_orders.to_sql('Orders', conn, if_exists='replace', index=False)

print("✅ Tables split and saved successfully!")

conn.close()
```
